Data Engineering Context:
Virtual environments isolate project dependencies,
preventing conflicts between different pipelines
and ensuring reproducible environments.


```python
import sys
import os
from pathlib import Path
```

## Why Virtual Environments?


```python
print("=== Why Virtual Environments? ===")

print("""
PROBLEM: Dependency conflicts between projects

Project A needs:        Project B needs:
  pandas==1.5.0          pandas==2.0.0
  numpy==1.23.0          numpy==1.24.0

Without virtual environments:
  - Installing Project B breaks Project A
  - Global Python gets cluttered
  - "It works on my machine" problems

SOLUTION: Virtual environments
  - Isolated Python installation per project
  - Each project has its own dependencies
  - Reproducible across machines
""")
```

## Creating Virtual Environments


```python
print("\n=== Creating Virtual Environments ===")

print("""
USING venv (built into Python 3.3+):

# Create virtual environment
python -m venv venv

# Alternative names:
python -m venv .venv     # Hidden directory
python -m venv env       # Shorter name

PROJECT STRUCTURE:
my_project/
├── venv/                # Virtual environment (don't commit!)
│   ├── bin/             # Scripts (Linux/Mac)
│   ├── Scripts/         # Scripts (Windows)
│   ├── lib/             # Installed packages
│   └── pyvenv.cfg       # Configuration
├── src/
├── tests/
├── requirements.txt
└── .gitignore           # Should include venv/
""")
```

## Activating Virtual Environments


```python
print("\n=== Activating Virtual Environments ===")

print("""
ACTIVATION COMMANDS:

Linux/Mac (bash/zsh):
  source venv/bin/activate

Windows (Command Prompt):
  venv\\Scripts\\activate.bat

Windows (PowerShell):
  venv\\Scripts\\Activate.ps1

AFTER ACTIVATION:
  - Prompt shows (venv) prefix
  - python points to venv Python
  - pip installs to venv only

DEACTIVATION:
  deactivate
""")

# Show current Python info
print(f"\nCurrent Python executable: {sys.executable}")
print(f"Python version: {sys.version}")

# Check if in virtual environment
in_venv = sys.prefix != sys.base_prefix
print(f"In virtual environment: {in_venv}")
```

## Managing Packages With Pip


```python
print("\n=== Managing Packages with pip ===")

print("""
BASIC COMMANDS:

# Install package
pip install pandas

# Install specific version
pip install pandas==2.0.0

# Install with version constraints
pip install "pandas>=1.5,<2.0"

# Install from requirements file
pip install -r requirements.txt

# Upgrade package
pip install --upgrade pandas

# Uninstall package
pip uninstall pandas

# List installed packages
pip list

# Show package info
pip show pandas

# Check for outdated packages
pip list --outdated
""")
```

## Requirements.Txt


```python
print("\n=== requirements.txt ===")

print("""
CREATING requirements.txt:

# Method 1: Freeze current environment (ALL packages)
pip freeze > requirements.txt
```

Method 2: Manual curation (RECOMMENDED)
Create requirements.txt manually with direct dependencies only

```python
EXAMPLE requirements.txt:
# Core dependencies
pandas>=2.0.0,<3.0.0
numpy>=1.24.0
sqlalchemy>=2.0.0

# Testing
pytest>=7.0.0
pytest-cov>=4.0.0
```

Optional: Development tools
black>=23.0.0
mypy>=1.0.0

```python
VERSION SPECIFIERS:
  pandas==2.0.0      # Exact version
  pandas>=2.0.0      # Minimum version
  pandas<3.0.0       # Maximum version
  pandas>=2.0,<3.0   # Version range
  pandas~=2.0.0      # Compatible release (~=2.0.0 means >=2.0.0,<2.1.0)
""")
```

## Requirements File Organization


```python
print("\n=== Requirements File Organization ===")

print("""
MULTIPLE REQUIREMENTS FILES:

requirements/
├── base.txt          # Core dependencies
├── dev.txt           # Development tools
├── test.txt          # Testing dependencies
└── prod.txt          # Production-specific

# base.txt
pandas>=2.0.0
sqlalchemy>=2.0.0

# dev.txt
-r base.txt           # Include base
black>=23.0.0
mypy>=1.0.0
ipython>=8.0.0

# test.txt
-r base.txt           # Include base
pytest>=7.0.0
pytest-cov>=4.0.0

# prod.txt
-r base.txt           # Include base
gunicorn>=21.0.0

USAGE:
pip install -r requirements/dev.txt   # Install dev dependencies
pip install -r requirements/prod.txt  # Install prod dependencies
""")
```

## Pinning Dependencies


```python
print("\n=== Pinning Dependencies ===")

print("""
LOOSE PINNING (development):
  pandas>=2.0.0
  # Allows updates, might break

STRICT PINNING (production):
  pandas==2.0.3
  numpy==1.24.2
  # Exact versions, reproducible

BEST PRACTICE:
  1. requirements.txt - loose pins (direct dependencies)
  2. requirements-lock.txt - strict pins (all dependencies)

GENERATE LOCK FILE:
  pip install -r requirements.txt
  pip freeze > requirements-lock.txt

PRODUCTION DEPLOY:
  pip install -r requirements-lock.txt
""")
```

## Common Pip Options


```python
print("\n=== Common pip Options ===")

print("""
USEFUL OPTIONS:

# Install without cache (saves space)
pip install --no-cache-dir pandas

# Install to user directory (no venv)
pip install --user pandas

# Install editable (for development)
pip install -e .

# Install from git
pip install git+https://github.com/user/repo.git

# Install from local directory
pip install ./my_package/

# Quiet mode (less output)
pip install -q pandas

# Verbose mode (more output)
pip install -v pandas

# Dry run (show what would be installed)
pip install --dry-run pandas
""")
```

## Alternative Tools


```python
print("\n=== Alternative Tools ===")

print("""
BEYOND venv + pip:

1. PIPENV
   - Combines venv + pip
   - Uses Pipfile instead of requirements.txt
   - Automatic lock file
   pipenv install pandas
   pipenv shell

2. POETRY
   - Modern dependency management
   - Uses pyproject.toml
   - Built-in packaging support
   poetry add pandas
   poetry shell

3. CONDA
   - From Anaconda ecosystem
   - Manages Python itself + packages
   - Great for data science (numpy, scipy)
   conda create -n myenv python=3.11
   conda activate myenv
   conda install pandas

4. UV (new, fast)
   - Rust-based, very fast
   - Drop-in pip replacement
   uv venv
   uv pip install pandas

RECOMMENDATION FOR DATA ENGINEERING:
  - Start with venv + pip (standard, simple)
  - Consider poetry for larger projects
  - Use conda for complex scientific dependencies
""")
```

## Practical Workflow


```python
print("\n=== Practical Workflow ===")

print("""
NEW PROJECT SETUP:

# 1. Create project directory
mkdir my_etl_pipeline
cd my_etl_pipeline

# 2. Create virtual environment
python -m venv venv

# 3. Activate it
source venv/bin/activate  # Linux/Mac

# 4. Upgrade pip
pip install --upgrade pip

# 5. Install dependencies
pip install pandas sqlalchemy pytest

# 6. Save requirements
pip freeze > requirements.txt

# 7. Add venv to .gitignore
echo "venv/" >> .gitignore

# 8. Commit
git add .
git commit -m "Initial project setup"


CLONING EXISTING PROJECT:

# 1. Clone repository
git clone https://github.com/user/project.git
cd project

# 2. Create virtual environment
python -m venv venv

# 3. Activate it
source venv/bin/activate

# 4. Install dependencies
pip install -r requirements.txt

# 5. Run tests to verify
pytest
""")
```

## Troubleshooting


```python
print("\n=== Troubleshooting ===")

print("""
COMMON ISSUES:

1. "pip not found" after activation
   # Use python -m pip instead
   python -m pip install pandas

2. Permission denied
```

Don't use sudo with venv!
Make sure venv is activated

```python
3. Wrong Python version
   # Specify version when creating venv
   python3.11 -m venv venv

4. Package conflicts
   # Create fresh venv
   rm -rf venv
   python -m venv venv
   pip install -r requirements.txt

5. "Requirement already satisfied" but wrong version
   pip install --force-reinstall pandas==2.0.0

6. SSL certificate errors
   pip install --trusted-host pypi.org pandas
""")
```

## Summary


```python
print("\n=== Summary ===")
print("""
Virtual Environments Key Points:

CREATING:
  python -m venv venv

ACTIVATING:
  source venv/bin/activate  # Linux/Mac
  venv\\Scripts\\activate    # Windows

DEACTIVATING:
  deactivate

REQUIREMENTS:
  pip freeze > requirements.txt     # Export
  pip install -r requirements.txt   # Import

BEST PRACTICES:
  - One venv per project
  - Don't commit venv/ to git
  - Use requirements.txt for dependencies
  - Pin versions for reproducibility
  - Separate dev/test/prod requirements

ALTERNATIVES:
  - pipenv: Pipfile + lock file
  - poetry: Modern, uses pyproject.toml
  - conda: Great for data science
  - uv: Fast, Rust-based
""")
```
