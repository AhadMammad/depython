## String Creation


```python
print("=== String Creation ===")

# Single and double quotes are interchangeable
table_name = 'customers'
schema_name = "staging"
print(f"Table: {schema_name}.{table_name}")

# Use one type to include the other without escaping
query = "SELECT * FROM 'customers'"  # Single quotes inside double
message = 'He said "Hello"'  # Double quotes inside single
print(query)
print(message)

# Escape quotes when needed
escaped = 'It\'s a table called "customers"'
print(escaped)

# Triple quotes for multi-line strings
sql_query = """
SELECT
    customer_id,
    customer_name,
    created_at
FROM staging.customers
WHERE status = 'active'
"""
print(f"SQL Query:{sql_query}")
```

## F-Strings (Formatted String Literals) — Python 3.6+


```python
print("\n=== F-Strings ===")

# Basic f-string
pipeline_name = "daily_etl"
status = "completed"
print(f"Pipeline '{pipeline_name}' {status}")

# Expressions inside f-strings
records_processed = 1500
records_failed = 23
print(f"Success rate: {(records_processed - records_failed) / records_processed:.2%}")
# OUTPUT: Success rate: 98.47%

# Formatting numbers
large_number = 1234567890
print(f"With commas: {large_number:,}")        # OUTPUT: 1,234,567,890
print(f"Scientific: {large_number:.2e}")       # OUTPUT: 1.23e+09
print(f"Padded: {42:05d}")                     # OUTPUT: 00042

# Formatting floats
pi = 3.14159265359
print(f"Two decimals: {pi:.2f}")               # OUTPUT: 3.14
print(f"Four decimals: {pi:.4f}")              # OUTPUT: 3.1416

# Alignment and width
name = "ETL"
print(f"Left aligned:  '{name:<10}'")          # OUTPUT: 'ETL       '
print(f"Right aligned: '{name:>10}'")          # OUTPUT: '       ETL'
print(f"Centered:      '{name:^10}'")          # OUTPUT: '   ETL    '

# Debug format (Python 3.8+) — shows variable name and value
batch_size = 100
print(f"{batch_size=}")  # OUTPUT: batch_size=100

# Calling methods inside f-strings
table = "  RAW_CUSTOMERS  "
print(f"Cleaned: {table.strip().lower()}")  # OUTPUT: raw_customers
```

## Essential String Methods


```python
print("\n=== Essential String Methods ===")

# Sample data
raw_column = "  Customer Name  "
file_path = "/data/raw/customers_2024_01_15.csv"
log_line = "ERROR:pipeline:Connection failed:Timeout"

# --- Whitespace handling ---
print("--- strip(), lstrip(), rstrip() ---")
print(f"Original: '{raw_column}'")
print(f"strip():  '{raw_column.strip()}'")     # Removes both ends
print(f"lstrip(): '{raw_column.lstrip()}'")    # Removes left
print(f"rstrip(): '{raw_column.rstrip()}'")    # Removes right

# --- Case conversion ---
print("\n--- lower(), upper(), title(), capitalize() ---")
text = "customer NAME"
print(f"lower():      '{text.lower()}'")       # OUTPUT: customer name
print(f"upper():      '{text.upper()}'")       # OUTPUT: CUSTOMER NAME
print(f"title():      '{text.title()}'")       # OUTPUT: Customer Name
print(f"capitalize(): '{text.capitalize()}'")  # OUTPUT: Customer name

# --- Splitting and joining ---
print("\n--- split() and join() ---")
print(f"Original: '{log_line}'")
parts = log_line.split(":")
print(f"split(':'): {parts}")
# OUTPUT: ['ERROR', 'pipeline', 'Connection failed', 'Timeout']

# Limit splits
parts_limited = log_line.split(":", maxsplit=2)
print(f"split(':', maxsplit=2): {parts_limited}")
# OUTPUT: ['ERROR', 'pipeline', 'Connection failed:Timeout']

# Join — opposite of split
columns = ["id", "name", "email", "created_at"]
column_list = ", ".join(columns)
print(f"join(): {column_list}")
# OUTPUT: id, name, email, created_at

# Split by whitespace (default)
text = "one   two  three"
print(f"'{text}'.split(): {text.split()}")  # OUTPUT: ['one', 'two', 'three']

# --- Search and check ---
print("\n--- find(), index(), count(), startswith(), endswith() ---")

# find() returns -1 if not found
print(f"'{file_path}'.find('csv'): {file_path.find('csv')}")        # OUTPUT: 30
print(f"'{file_path}'.find('parquet'): {file_path.find('parquet')}") # OUTPUT: -1
```

index() raises ValueError if not found (use find() for safe search)
count() counts occurrences

```python
text = "data_data_pipeline_data"
print(f"count('data'): {text.count('data')}")  # OUTPUT: 3

# startswith() and endswith()
print(f"startswith('/data'): {file_path.startswith('/data')}")  # OUTPUT: True
print(f"endswith('.csv'): {file_path.endswith('.csv')}")        # OUTPUT: True
print(f"endswith(('.csv', '.parquet')): {file_path.endswith(('.csv', '.parquet'))}")

# --- Replace ---
print("\n--- replace() ---")
query = "SELECT * FROM {table} WHERE {condition}"
query = query.replace("{table}", "customers")
query = query.replace("{condition}", "active = true")
print(f"Replaced: {query}")

# Clean column names
dirty_column = "Customer First Name (Primary)"
clean_column = dirty_column.lower().replace(" ", "_").replace("(", "").replace(")", "")
print(f"Cleaned column: {clean_column}")  # OUTPUT: customer_first_name_primary
```

## String Slicing


```python
print("\n=== String Slicing ===")
```

Syntax: string[start:stop:step]
start: inclusive, stop: exclusive

```python
text = "ABCDEFGHIJ"
print(f"text = '{text}'")
print(f"text[0]:     '{text[0]}'")      # First char: A
print(f"text[-1]:    '{text[-1]}'")     # Last char: J
print(f"text[0:3]:   '{text[0:3]}'")    # First 3: ABC
print(f"text[:3]:    '{text[:3]}'")     # Same as above: ABC
print(f"text[3:]:    '{text[3:]}'")     # From index 3 to end: DEFGHIJ
print(f"text[-3:]:   '{text[-3:]}'")    # Last 3: HIJ
print(f"text[::2]:   '{text[::2]}'")    # Every 2nd: ACEGI
print(f"text[::-1]:  '{text[::-1]}'")   # Reversed: JIHGFEDCBA

# Practical: Extract parts from file path
file_path = "/data/warehouse/customers_2024_01_15.csv"

# Get filename
filename = file_path.split("/")[-1]
print(f"\nFilename: {filename}")  # OUTPUT: customers_2024_01_15.csv

# Get extension
extension = filename.split(".")[-1]
print(f"Extension: {extension}")  # OUTPUT: csv

# Get date from filename (assuming pattern name_YYYY_MM_DD.ext)
date_part = filename[-14:-4]  # Extract YYYY_MM_DD
print(f"Date part: {date_part}")  # OUTPUT: 2024_01_15
```

## Regular Expressions Basics


```python
print("\n=== Regular Expressions ===")
```

Import: import re (at top of file)
--- re.search() — Find first match anywhere in string ---

```python
text = "Pipeline completed at 2024-01-15 14:30:00"
date_pattern = r"\d{4}-\d{2}-\d{2}"  # YYYY-MM-DD

match = re.search(date_pattern, text)
if match:
    print(f"Found date: {match.group()}")  # OUTPUT: 2024-01-15

# --- re.match() — Match at beginning of string only ---
log_line = "ERROR: Connection failed"
error_pattern = r"ERROR:"

if re.match(error_pattern, log_line):
    print("Line starts with ERROR")

# --- re.findall() — Find all matches ---
text = "IDs: 101, 202, 303, 404"
id_pattern = r"\d+"

ids = re.findall(id_pattern, text)
print(f"All IDs: {ids}")  # OUTPUT: ['101', '202', '303', '404']

# Convert to integers
ids_int = [int(id) for id in ids]
print(f"As integers: {ids_int}")  # OUTPUT: [101, 202, 303, 404]

# --- re.sub() — Replace using patterns ---
text = "customer_id,customer_name,customer_email"
# Remove "customer_" prefix from all columns
cleaned = re.sub(r"customer_", "", text)
print(f"Cleaned: {cleaned}")  # OUTPUT: id,name,email

# Complex replacement: standardize phone formats
phones = "Call 123-456-7890 or 987.654.3210 or (555) 123 4567"
# Replace various separators with dashes
standardized = re.sub(r"[\s.\-()]+", "-", phones)
standardized = re.sub(r"-+", "-", standardized)  # Remove multiple dashes
print(f"Standardized: {standardized}")

# --- Capturing groups ---
log = "2024-01-15 ERROR pipeline_daily: Connection timeout"
pattern = r"(\d{4}-\d{2}-\d{2}) (\w+) (\w+): (.+)"

match = re.match(pattern, log)
if match:
    date, level, pipeline, message = match.groups()
    print(f"\nParsed log:")
    print(f"  Date: {date}")
    print(f"  Level: {level}")
    print(f"  Pipeline: {pipeline}")
    print(f"  Message: {message}")
```

## Practical: Cleaning Column Names


```python
print("\n=== Practical: Cleaning Column Names ===")

def clean_column_name(name: str) -> str:
    """
    Clean a column name for database compatibility.

    - Converts to lowercase
    - Replaces spaces with underscores
    - Removes special characters
    - Removes leading/trailing underscores
    """
    # Convert to lowercase
    cleaned = name.lower()

    # Replace spaces and common separators with underscores
    cleaned = re.sub(r"[\s\-]+", "_", cleaned)

    # Remove special characters (keep only alphanumeric and underscores)
    cleaned = re.sub(r"[^a-z0-9_]", "", cleaned)

    # Remove multiple consecutive underscores
    cleaned = re.sub(r"_+", "_", cleaned)

    # Remove leading/trailing underscores
    cleaned = cleaned.strip("_")

    return cleaned


# Test with messy column names
messy_columns = [
    "Customer ID",
    "First Name (Legal)",
    "E-mail Address",
    "  Phone #  ",
    "Created_At__UTC",
]

print("Original → Cleaned:")
for col in messy_columns:
    print(f"  '{col}' → '{clean_column_name(col)}'")

# OUTPUT:
```

'Customer ID' → 'customer_id'
'First Name (Legal)' → 'first_name_legal'
'E-mail Address' → 'email_address'
'  Phone #  ' → 'phone'
'Created_At__UTC' → 'created_at_utc'
