## For Loops — Iterate Over Sequences


```python
print("=== for Loops ===")

# Basic for loop over a list
tables = ["customers", "orders", "products", "inventory"]

print("Tables to process:")
for table in tables:
    print(f"  - {table}")

# For loop with range() — generate sequence of numbers
print("\nProcessing 5 batches:")
for batch_num in range(5):  # 0, 1, 2, 3, 4
    print(f"  Batch {batch_num}")

# range(start, stop) — stop is exclusive
print("\nRecord IDs 100-104:")
for record_id in range(100, 105):  # 100, 101, 102, 103, 104
    print(f"  ID: {record_id}")

# range(start, stop, step) — with step
print("\nEvery 10th record (10-50):")
for record_id in range(10, 51, 10):  # 10, 20, 30, 40, 50
    print(f"  ID: {record_id}")

# Counting backwards
print("\nCountdown:")
for i in range(5, 0, -1):  # 5, 4, 3, 2, 1
    print(f"  {i}...")
print("  Liftoff!")
```

## Enumerate() — Get Index And Value


```python
print("\n=== enumerate() ===")

columns = ["customer_id", "name", "email", "created_at"]

# Without enumerate (not Pythonic)
print("Without enumerate:")
for i in range(len(columns)):
    print(f"  Column {i}: {columns[i]}")

# With enumerate (Pythonic!)
print("\nWith enumerate:")
for index, column in enumerate(columns):
    print(f"  Column {index}: {column}")

# Start from different index
print("\nWith enumerate(start=1):")
for position, column in enumerate(columns, start=1):
    print(f"  Position {position}: {column}")

# Practical: Number lines in a log file
log_lines = [
    "INFO: Pipeline started",
    "DEBUG: Loading config",
    "INFO: Processing records",
    "ERROR: Connection failed",
]

print("\nLog with line numbers:")
for line_num, line in enumerate(log_lines, start=1):
    print(f"  {line_num:3}: {line}")
```

## Zip() — Iterate Over Multiple Sequences Together


```python
print("\n=== zip() ===")

# Pair up elements from two lists
column_names = ["id", "name", "price", "quantity"]
column_types = ["INTEGER", "VARCHAR", "DECIMAL", "INTEGER"]

print("Column definitions:")
for name, dtype in zip(column_names, column_types):
    print(f"  {name}: {dtype}")

# Zip with three lists
column_names = ["id", "name", "email"]
old_values = [1, "Alice", "alice@old.com"]
new_values = [1, "Alice Smith", "alice@new.com"]

print("\nDetecting changes:")
for col, old, new in zip(column_names, old_values, new_values):
    if old != new:
        print(f"  {col}: '{old}' → '{new}'")

# Zip stops at shortest sequence
short_list = [1, 2]
long_list = [10, 20, 30, 40]
print(f"\nzip({short_list}, {long_list}):")
for a, b in zip(short_list, long_list):
    print(f"  {a}, {b}")
```

Only prints (1, 10) and (2, 20)
Use zip_longest for full iteration (from itertools)

```python
from itertools import zip_longest

print(f"\nzip_longest({short_list}, {long_list}, fillvalue=0):")
for a, b in zip_longest(short_list, long_list, fillvalue=0):
    print(f"  {a}, {b}")
```

## While Loops — Conditional Iteration


```python
print("\n=== while Loops ===")

# Basic while loop
retry_count = 0
max_retries = 3

print("Retry simulation:")
while retry_count < max_retries:
    print(f"  Attempt {retry_count + 1}")
    retry_count += 1
print("  Max retries reached")

# While with condition that changes
print("\nProcessing queue:")
queue = ["task1", "task2", "task3"]
while queue:  # While queue is not empty
    task = queue.pop(0)  # Remove first item
    print(f"  Processing: {task}")
print("  Queue empty")

# Practical: Read until sentinel value
print("\nReading until empty record:")
records = [
    {"id": 1, "name": "Alice"},
    {"id": 2, "name": "Bob"},
    {},  # Empty dict signals end
    {"id": 3, "name": "Charlie"},  # Won't be processed
]

index = 0
while index < len(records) and records[index]:
    print(f"  Processing: {records[index]}")
    index += 1
print("  Stopped at empty record")
```

## Break, Continue, And Else


```python
print("\n=== break, continue, else ===")

# break — Exit loop immediately
print("Finding first error:")
log_lines = ["INFO: Start", "DEBUG: Config", "ERROR: Failure", "INFO: End"]
for line in log_lines:
    if "ERROR" in line:
        print(f"  Found error: {line}")
        break
    print(f"  Checked: {line}")

# continue — Skip to next iteration
print("\nSkipping DEBUG lines:")
for line in log_lines:
    if "DEBUG" in line:
        continue  # Skip DEBUG lines
    print(f"  {line}")

# else clause on loops — Executes if loop completes WITHOUT break
print("\nSearching for 'CRITICAL':")
for line in log_lines:
    if "CRITICAL" in line:
        print(f"  Found: {line}")
        break
else:
    # This runs because we didn't break
    print("  No CRITICAL errors found")

print("\nSearching for 'ERROR':")
for line in log_lines:
    if "ERROR" in line:
        print(f"  Found: {line}")
        break
else:
    # This does NOT run because we broke
    print("  No ERROR found")

# Practical: Validate all records
print("\nValidating records:")
records = [
    {"id": 1, "status": "active"},
    {"id": 2, "status": "active"},
    {"id": 3, "status": "active"},
]

for record in records:
    if record["status"] != "active":
        print(f"  Invalid record found: {record['id']}")
        break
else:
    print("  All records are valid ✓")
```

## Iterating Over Dictionaries


```python
print("\n=== Iterating Over Dictionaries ===")

pipeline_config = {
    "name": "daily_etl",
    "batch_size": 1000,
    "timeout": 300,
    "enabled": True,
}

# Iterate over keys (default)
print("Keys:")
for key in pipeline_config:
    print(f"  {key}")

# Iterate over values
print("\nValues:")
for value in pipeline_config.values():
    print(f"  {value}")

# Iterate over key-value pairs (most common)
print("\nKey-Value pairs:")
for key, value in pipeline_config.items():
    print(f"  {key}: {value}")

# Practical: Filter config by type
print("\nString config values:")
for key, value in pipeline_config.items():
    if isinstance(value, str):
        print(f"  {key}: {value}")
```

## Nested Loops


```python
print("\n=== Nested Loops ===")

# Process tables and their columns
schemas = {
    "customers": ["id", "name", "email"],
    "orders": ["id", "customer_id", "total"],
}

print("Schema structure:")
for table_name, columns in schemas.items():
    print(f"\n  Table: {table_name}")
    for column in columns:
        print(f"    - {column}")

# Create column mappings across tables
print("\nGenerating column mappings:")
source_columns = ["first_name", "last_name", "email_address"]
target_columns = ["fname", "lname", "email"]

for source, target in zip(source_columns, target_columns):
    print(f"  {source} → {target}")

# Nested loops with indices
print("\nProcessing batches of records:")
num_batches = 3
records_per_batch = 2

for batch in range(num_batches):
    print(f"  Batch {batch + 1}:")
    for record in range(records_per_batch):
        record_id = batch * records_per_batch + record + 1
        print(f"    Record {record_id}")
```

## Practical: Batch Processing With Retry


```python
print("\n=== Practical: Batch Processing with Retry ===")

def process_batch(batch_id: int, fail_on: int = -1) -> bool:
    """Simulate batch processing. Fails if batch_id == fail_on."""
    if batch_id == fail_on:
        return False
    return True


batches = [1, 2, 3, 4, 5]
max_retries = 3
failed_batches = []

print("Processing batches with retry logic:")
for batch_id in batches:
    retry = 0
    success = False

    while retry < max_retries and not success:
        if retry > 0:
            print(f"    Retry {retry} for batch {batch_id}")

        # Simulate: batch 3 fails first 2 attempts, succeeds on 3rd
        fail_threshold = 2 if batch_id == 3 else -1
        success = process_batch(batch_id, fail_on=(batch_id if retry < fail_threshold else -1))

        if not success:
            retry += 1

    if success:
        print(f"  ✓ Batch {batch_id} processed")
    else:
        print(f"  ✗ Batch {batch_id} failed after {max_retries} retries")
        failed_batches.append(batch_id)

print(f"\nFailed batches: {failed_batches if failed_batches else 'None'}")
```
