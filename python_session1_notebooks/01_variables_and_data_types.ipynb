## Numeric Types: Int And Float


Integer — whole numbers, no decimal point
Common in data engineering: row counts, batch sizes, retry counts

```python
row_count = 1000000000  # Underscores improve readability for large numbers
batch_size = 500
max_retries = 3

print("=== Numeric Types ===")
print(f"Row count: {row_count}")  # OUTPUT: Row count: 1000000
print(f"Batch size: {batch_size}")  # OUTPUT: Batch size: 500
print(f"Type of row_count: {type(row_count)}")  # OUTPUT: Type of row_count: <class 'int'>
```

Float — decimal numbers
Common in data engineering: percentages, metrics, thresholds

```python
success_rate = 0.95
memory_usage_gb = 3.75
threshold = 0.001

print(f"\nSuccess rate: {success_rate}")  # OUTPUT: Success rate: 0.95
print(f"Type of success_rate: {type(success_rate)}")  # OUTPUT: Type of success_rate: <class 'float'>

# Be aware of floating point precision issues
print(f"\n0.1 + 0.9 = {0.1 + 0.2}")  # OUTPUT: 0.1 + 0.2 = 0.30000000000000004
# For financial/precise calculations, use the decimal module
```

## String Type: Str


Strings — text data, immutable sequences of characters
Common in data engineering: table names, file paths, SQL queries

```python
table_name = "raw_customers"
schema_name = 'staging'  # Single or double quotes both work
file_path = "/data/warehouse/customers.parquet"

print("\n=== String Type ===")
print(f"Table: {schema_name}.{table_name}")  # OUTPUT: Table: staging.raw_customers
print(f"Type of table_name: {type(table_name)}")  # OUTPUT: Type of table_name: <class 'str'>
```

Multi-line strings with triple quotes
Great for SQL queries and documentation

```python
sql_query = """
SELECT customer_id,
       customer_name,
       created_at
FROM staging.raw_customers
WHERE active = false
"""
print(f"SQL Query:{sql_query}")
```

## Boolean Type: Bool


Booleans — True or False
Common in data engineering: flags, conditions, status checks

```python
is_production = True
has_header = True
skip_validation = False
pipeline_enabled = True

print("\n=== Boolean Type ===")
print(f"Is production: {is_production}")  # OUTPUT: Is production: True
print(f"Type of is_production: {type(is_production)}")  # OUTPUT: Type of is_production: <class 'bool'>

# Booleans are actually integers under the hood
print(f"True + True = {True + True}")  # OUTPUT: True + True = 2
print(f"True * 10 = {True * 10}")  # OUTPUT: True * 10 = 10
```

## Complex Type: Complex (Rarely Used In Data Engineering)


```python
# Complex numbers — real + imaginary parts
complex_number = 3 + 4j

print("\n=== Complex Type ===")
print(f"Complex number: {complex_number}")  # OUTPUT: Complex number: (3+4j)
print(f"Real part: {complex_number.real}")  # OUTPUT: Real part: 3.0
print(f"Imaginary part: {complex_number.imag}")  # OUTPUT: Imaginary part: 4.0
```

============================================================
type() FUNCTION — Check the type of any value
============================================================

```python
print("\n=== Using type() ===")

# type() returns the class of an object
pipeline_name = "etl_daily_sales"
record_count = 42
is_active = True

print(f"type('{pipeline_name}') = {type(pipeline_name)}")
# OUTPUT: type('etl_daily_sales') = <class 'str'>

print(f"type({record_count}) = {type(record_count)}")
# OUTPUT: type(42) = <class 'int'>

print(f"type({is_active}) = {type(is_active)}")
# OUTPUT: type(True) = <class 'bool'>

# Comparing types directly
print(f"\ntype(record_count) == int: {type(record_count) == int}")  # OUTPUT: True
```

============================================================
isinstance() FUNCTION — Check if object is instance of a type
============================================================

```python
print("\n=== Using isinstance() ===")
```

isinstance() is preferred over type() for type checking
because it handles inheritance properly

```python
value = 100

print(f"isinstance({value}, int): {isinstance(value, int)}")  # OUTPUT: True
print(f"isinstance({value}, float): {isinstance(value, float)}")  # OUTPUT: False
print(f"isinstance({value}, (int, float)): {isinstance(value, (int, float))}")  # OUTPUT: True


size = 100
print(f"{size}")

# Practical example: validating configuration values
def validate_batch_size(size):
    """Validate that batch size is a positive integer."""
    if not isinstance(size, int):
        raise TypeError(f"Batch size must be int, got {type(size).__name__}")
    if size <= 0:
        raise ValueError("Batch size must be positive")
    return True

# This will work
print(f"\nvalidate_batch_size(100): {validate_batch_size(100)}")  # OUTPUT: True
```

## Dynamic Typing — Variables Can Change Type


```python
print("\n=== Dynamic Typing ===")
```

Python doesn't require type declarations
Variables can be reassigned to different types

```python
status = "pending"  # str
print(f"status = '{status}', type: {type(status).__name__}")

status = 1  # Now it's an int
print(f"status = {status}, type: {type(status).__name__}")

status = True  # Now it's a bool
print(f"status = {status}, type: {type(status).__name__}")
```

This flexibility is powerful but requires discipline
Use type hints to document expected types (covered in module 08)

## Mutable Vs Immutable — Introduction


```python
print("\n=== Mutable vs Immutable ===")
```

IMMUTABLE types: int, float, str, bool, tuple, frozenset
You cannot change them in place; reassignment creates a new object

```python
original_count = 10
print(f"original_count id: {id(original_count)}")

original_count = 20  # This creates a NEW int object
print(f"After reassignment, id: {id(original_count)}")  # Different id!

# Strings are immutable
name = "pipeline"
print(f"\nString '{name}' id: {id(name)}")
name = name.upper()  # Creates a NEW string
print(f"After upper(), '{name}' id: {id(name)}")  # Different id!
```

MUTABLE types: list, dict, set
You CAN change them in place (covered in Session 2)

```python
columns = ["id", "name", "email"]
print(f"\nList {columns} id: {id(columns)}")
columns.append("created_at")  # Modifies in place
print(f"After append(), {columns} id: {id(columns)}")  # SAME id!
```

Why does this matter?
- Immutable types are hashable (can be dict keys)
- Mutable types can cause unexpected side effects if shared

## Practical Example: Pipeline Configuration


```python
print("\n=== Practical Example: Pipeline Config ===")

# A typical pipeline might have these configuration values
pipeline_config = {
    "name": "daily_sales_etl",        # str
    "batch_size": 10000,               # int
    "memory_limit_gb": 4.5,            # float
    "is_production": True,             # bool
    "source_tables": ["orders", "customers", "products"],  # list of str
}

# Access and display configuration
print(f"Pipeline: {pipeline_config['name']}")
print(f"Batch size: {pipeline_config['batch_size']:,} records")
print(f"Memory limit: {pipeline_config['memory_limit_gb']} GB")
print(f"Production mode: {pipeline_config['is_production']}")
print(f"Source tables: {', '.join(pipeline_config['source_tables'])}")

# Type check a config value
if isinstance(pipeline_config['batch_size'], int):
    print("\n✓ Batch size configuration is valid")
```
