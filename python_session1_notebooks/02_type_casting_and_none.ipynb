## Type Casting — Converting Between Types


```python
print("=== Type Casting ===")

# int() — Convert to integer
string_count = "1000"
int_count = int(string_count)
print(f"int('{string_count}') = {int_count}, type: {type(int_count).__name__}")
# OUTPUT: int('1000') = 1000, type: int

# int() truncates floats (doesn't round!)
float_value = 3.9
print(f"int({float_value}) = {int(float_value)}")  # OUTPUT: int(3.9) = 3

# float() — Convert to float
string_price = "19.99"
float_price = float(string_price)
print(f"\nfloat('{string_price}') = {float_price}")  # OUTPUT: float('19.99') = 19.99

int_value = 42
print(f"float({int_value}) = {float(int_value)}")  # OUTPUT: float(42) = 42.0

# str() — Convert to string
record_count = 50000
str_count = str(record_count)
print(f"\nstr({record_count}) = '{str_count}', type: {type(str_count).__name__}")
# OUTPUT: str(50000) = '50000', type: str

# Common pattern: building log messages or SQL
table_name = "customers"
count = 1500
log_message = "Processed " + str(count) + " records from " + table_name
print(f"Log: {log_message}")

# bool() — Convert to boolean
print(f"\nbool(1) = {bool(1)}")        # OUTPUT: True
print(f"bool(0) = {bool(0)}")          # OUTPUT: False
print(f"bool('') = {bool('')}")        # OUTPUT: False (empty string)
print(f"bool('text') = {bool('text')}") # OUTPUT: True
```

## Handling Conversion Errors


```python
print("\n=== Handling Conversion Errors ===")

# Converting invalid strings raises ValueError
raw_value = "not_a_number"

try:
    number = int(raw_value)
except ValueError as e:
    print(f"Cannot convert '{raw_value}' to int: {e}")
    # OUTPUT: Cannot convert 'not_a_number' to int: invalid literal for int()...

# Safe conversion pattern for data pipelines
def safe_int(value, default=0):
    """Safely convert to int, return default if conversion fails."""
    try:
        return int(value)
    except (ValueError, TypeError):
        return default

print(f"\nsafe_int('100') = {safe_int('100')}")       # OUTPUT: 100
print(f"safe_int('abc') = {safe_int('abc')}")         # OUTPUT: 0
print(f"safe_int('abc', -1) = {safe_int('abc', -1)}") # OUTPUT: -1
```

## None — Python'S Null Value


```python
print("\n=== None — Python's Null ===")
```

None represents the absence of a value
Similar to NULL in databases, null in JSON, nil in other languages

```python
customer_middle_name = None  # Customer didn't provide middle name
last_processed_date = None   # Pipeline hasn't run yet

print(f"customer_middle_name = {customer_middle_name}")
print(f"Type of None: {type(None).__name__}")  # OUTPUT: <class 'NoneType'>

# None is a singleton — there's only one None object
a = None
b = None
print(f"\na is b: {a is b}")  # OUTPUT: True (same object)

# Functions return None implicitly if no return statement
def log_message(message):
    """Log a message (no explicit return)."""
    print(f"LOG: {message}")

result = log_message("Pipeline started")
print(f"Function returned: {result}")  # OUTPUT: None
```

## Checking For None — Use 'Is' Not '=='


```python
print("\n=== Checking for None ===")

value = None

# CORRECT: Use 'is' for None checks
if value is None:
    print("value is None (using 'is') ✓")

if value is not None:
    print("value is not None")
else:
    print("value is None (using 'is not') ✓")
```

Why 'is' and not '=='?
'is' checks identity (same object)
'==' checks equality (could be overridden by classes)

## Truthy And Falsy Values


```python
print("\n=== Truthy and Falsy Values ===")

# Python considers these values as "falsy" (evaluate to False in boolean context):
falsy_values = [
    False,      # Boolean False
    None,       # None
    0,          # Integer zero
    0.0,        # Float zero
    "",         # Empty string
    [],         # Empty list
    {},         # Empty dict
    set(),      # Empty set
    (),         # Empty tuple
]

print("Falsy values:")
for val in falsy_values:
    print(f"  bool({repr(val):12}) = {bool(val)}")

# Everything else is "truthy":
truthy_values = [
    True,
    1,
    -1,
    3.14,
    "text",
    [1, 2, 3],
    {"key": "value"},
    {1, 2, 3},
]

print("\nTruthy values:")
for val in truthy_values:
    print(f"  bool({repr(val):20}) = {bool(val)}")
```

## Common Gotcha: `If X` Vs `If X Is Not None`


```python
print("\n=== Common Gotcha: if x vs if x is not None ===")
```

These are NOT the same!
Scenario 1: Checking if a value exists

```python
row_count = 0  # Valid count — zero rows processed

# WRONG: This treats 0 as "no value"
if row_count:
    print(f"Processed {row_count} rows")
else:
    print("No row count provided (WRONG - we have 0!)")

# CORRECT: Explicitly check for None
if row_count is not None:
    print(f"Processed {row_count} rows (CORRECT)")

# Scenario 2: Optional string parameter
def greet(name=None):
    """Greet by name, or generically if no name provided."""
    if name is not None:
        return f"Hello, {name}!"
    return "Hello, guest!"

print(f"\ngreet('Alice') = {greet('Alice')}")
print(f"greet('') = {greet('')}")       # Empty string is still a value
print(f"greet() = {greet()}")           # None — no value provided
```

## Data Engineering Example: Processing Csv Values


```python
print("\n=== Data Engineering Example: CSV Processing ===")

# Raw CSV data often has empty strings or "NULL" text for missing values
raw_records = [
    {"name": "Alice", "age": "30", "middle_name": "Marie"},
    {"name": "Bob", "age": "", "middle_name": ""},       # Empty age
    {"name": "Charlie", "age": "25", "middle_name": "NULL"},  # NULL as text
]

def clean_value(value):
    """
    Clean a raw CSV value:
    - Empty string → None
    - "NULL" text → None
    - Otherwise, return stripped value
    """
    if value is None:
        return None

    stripped = value.strip()

    if stripped == "" or stripped.upper() == "NULL":
        return None

    return stripped

def parse_age(age_str):
    """Parse age string to int, return None if invalid."""
    cleaned = clean_value(age_str)
    if cleaned is None:
        return None
    try:
        return int(cleaned)
    except ValueError:
        return None

print("Cleaned records:")
for record in raw_records:
    cleaned_record = {
        "name": clean_value(record["name"]),
        "age": parse_age(record["age"]),
        "middle_name": clean_value(record["middle_name"]),
    }
    print(f"  {cleaned_record}")

# OUTPUT:
```
