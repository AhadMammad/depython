## List Creation And Basics


```python
print("=== List Creation and Basics ===")

# List creation
empty_list = []
columns = ["id", "name", "email", "created_at"]
mixed_types = [1, "two", 3.0, True, None]
nested_list = [[1, 2], [3, 4], [5, 6]]

print(f"Columns: {columns}")
print(f"Mixed types: {mixed_types}")
print(f"Nested: {nested_list}")

# Lists are mutable - can be changed in place
columns[0] = "customer_id"
print(f"Modified: {columns}")

# List length
print(f"Length: {len(columns)}")
```

## Indexing And Slicing


```python
print("\n=== Indexing and Slicing ===")

items = ["a", "b", "c", "d", "e", "f", "g", "h", "i", "j"]
print(f"items = {items}")

# Positive indexing (0-based)
print(f"items[0]:   '{items[0]}'")    # First element
print(f"items[3]:   '{items[3]}'")    # Fourth element

# Negative indexing (from the end)
print(f"items[-1]:  '{items[-1]}'")   # Last element
print(f"items[-3]:  '{items[-3]}'")   # Third from end

# Slicing: [start:stop:step]
print(f"\nitems[2:5]:   {items[2:5]}")     # Index 2, 3, 4
print(f"items[:3]:    {items[:3]}")        # First 3
print(f"items[-3:]:   {items[-3:]}")       # Last 3
print(f"items[::2]:   {items[::2]}")       # Every 2nd
print(f"items[1::2]:  {items[1::2]}")      # Every 2nd, starting at 1
print(f"items[::-1]:  {items[::-1]}")      # Reversed

# Practical: Get middle elements
data = list(range(10))
middle = data[len(data)//4 : 3*len(data)//4]
print(f"\nMiddle 50% of {data}: {middle}")
```

## List Methods


```python
print("\n=== List Methods ===")

# Starting list
tables = ["customers", "orders"]
print(f"Starting: {tables}")

# append() - Add single item to end
tables.append("products")
print(f"After append('products'): {tables}")

# extend() - Add multiple items
tables.extend(["inventory", "suppliers"])
print(f"After extend(['inventory', 'suppliers']): {tables}")

# insert() - Add at specific position
tables.insert(0, "schemas")  # Insert at beginning
print(f"After insert(0, 'schemas'): {tables}")

# remove() - Remove first occurrence by value
tables.remove("schemas")
print(f"After remove('schemas'): {tables}")

# pop() - Remove and return by index (default: last)
last = tables.pop()
print(f"pop() returned '{last}', list now: {tables}")

first = tables.pop(0)
print(f"pop(0) returned '{first}', list now: {tables}")

# index() - Find position of value
pos = tables.index("orders")
print(f"index('orders'): {pos}")

# count() - Count occurrences
items = [1, 2, 2, 3, 2, 4]
print(f"\n{items}.count(2): {items.count(2)}")

# sort() - Sort in place
numbers = [3, 1, 4, 1, 5, 9, 2, 6]
numbers.sort()
print(f"After sort(): {numbers}")

# Sort with key function
words = ["apple", "Banana", "cherry"]
words.sort(key=str.lower)  # Case-insensitive sort
print(f"After sort(key=str.lower): {words}")

# Sort descending
numbers.sort(reverse=True)
print(f"After sort(reverse=True): {numbers}")

# reverse() - Reverse in place
numbers.reverse()
print(f"After reverse(): {numbers}")
```

============================================================
sorted() vs .sort()
============================================================

```python
print("\n=== sorted() vs .sort() ===")

# .sort() modifies in place, returns None
original = [3, 1, 2]
result = original.sort()
print(f".sort() modifies in place, returns: {result}")
print(f"original is now: {original}")

# sorted() returns new list, original unchanged
original = [3, 1, 2]
result = sorted(original)
print(f"\nsorted() returns new list: {result}")
print(f"original unchanged: {original}")

# sorted() works on any iterable
text = "python"
print(f"sorted('{text}'): {sorted(text)}")
```

## Shallow Copy Vs Deep Copy


```python
print("\n=== Shallow Copy vs Deep Copy ===")

# Assignment creates a REFERENCE, not a copy
original = [1, 2, 3]
reference = original
reference.append(4)
print(f"original: {original}")  # [1, 2, 3, 4] - also modified!
print(f"reference: {reference}")  # [1, 2, 3, 4]

# Shallow copy - new list, but nested objects are references
original = [[1, 2], [3, 4]]
shallow = original.copy()  # or list(original) or original[:]

original.append([5, 6])  # Doesn't affect shallow
print(f"\noriginal after append: {original}")
print(f"shallow (unaffected): {shallow}")

original[0][0] = 999  # Affects both! Nested list is same object
print(f"original after modifying nested: {original}")
print(f"shallow (nested modified!): {shallow}")

# Deep copy - completely independent copy
original = [[1, 2], [3, 4]]
deep = copy.deepcopy(original)

original[0][0] = 999
print(f"\noriginal after modifying nested: {original}")
print(f"deep copy (unaffected): {deep}")

# Practical example: Copying records
records = [
    {"id": 1, "data": [1, 2, 3]},
    {"id": 2, "data": [4, 5, 6]},
]

# Shallow copy - records share nested data
shallow_records = records.copy()

# Deep copy - completely independent
deep_records = copy.deepcopy(records)

records[0]["data"].append(999)
print(f"\nOriginal records: {records}")
print(f"Shallow copy (nested affected): {shallow_records}")
print(f"Deep copy (independent): {deep_records}")
```

## List As Stack And Queue


```python
print("\n=== List as Stack (LIFO) ===")
```

Stack: Last In, First Out
Use append() to push, pop() to pop

```python
stack = []
stack.append("task1")  # Push
stack.append("task2")
stack.append("task3")
print(f"Stack: {stack}")

while stack:
    task = stack.pop()  # Pop from end (efficient)
    print(f"  Processing: {task}")
```

Queue: First In, First Out
WARNING: list is inefficient for queue (O(n) for pop(0))
Use collections.deque instead for real queues

```python
print("\n=== List as Queue (FIFO) ===")

from collections import deque

# Efficient queue with deque
queue = deque(["job1", "job2", "job3"])
print(f"Queue: {list(queue)}")

while queue:
    job = queue.popleft()  # O(1) from left
    print(f"  Processing: {job}")

# Add to queue
queue.append("job4")
queue.append("job5")
print(f"Queue after adding: {list(queue)}")
```

## Performance Characteristics


```python
print("\n=== Performance Characteristics ===")

print("""
List Operation Performance (Big O):
┌──────────────────────┬─────────────┐
│ Operation            │ Time        │
├──────────────────────┼─────────────┤
│ Index access [i]     │ O(1)        │
│ Append               │ O(1)*       │
│ Pop from end         │ O(1)        │
│ Pop from beginning   │ O(n)        │ ← Avoid!
│ Insert at beginning  │ O(n)        │ ← Avoid!
│ Insert at middle     │ O(n)        │
│ Search (in)          │ O(n)        │
│ Sort                 │ O(n log n)  │
│ Copy                 │ O(n)        │
└──────────────────────┴─────────────┘
* Amortized - occasional resize is O(n)

Recommendations:
- For queue operations, use collections.deque
- For membership testing, use set
- For sorted data with fast insert, consider bisect module
""")
```

## Practical: Batch Record Processing


```python
print("=== Practical: Batch Record Processing ===")


def process_records_in_batches(records: list, batch_size: int) -> list[list]:
    """
    Split records into batches for processing.

    Uses list slicing for efficient batching.
    """
    batches = []
    for i in range(0, len(records), batch_size):
        batch = records[i:i + batch_size]
        batches.append(batch)
    return batches


# Generate sample records
sample_records = [{"id": i, "value": i * 10} for i in range(1, 26)]

# Split into batches
batches = process_records_in_batches(sample_records, batch_size=10)

print(f"Total records: {len(sample_records)}")
print(f"Batch size: 10")
print(f"Number of batches: {len(batches)}")

for i, batch in enumerate(batches, 1):
    print(f"  Batch {i}: {len(batch)} records (IDs: {batch[0]['id']}-{batch[-1]['id']})")
```
