## Set Creation


```python
print("=== Set Creation ===")

# Using curly braces (note: {} is empty dict, not set!)
columns = {"id", "name", "email", "created_at"}
print(f"Set literal: {columns}")

# Using set() constructor
from_list = set([1, 2, 2, 3, 3, 3])  # Duplicates removed!
print(f"From list with duplicates: {from_list}")

# From string
from_string = set("hello")
print(f"From string: {from_string}")  # Unique characters

# Empty set (must use set(), not {})
empty_set = set()
print(f"Empty set: {empty_set}, type: {type(empty_set)}")

# Set comprehension
squares = {x ** 2 for x in range(1, 6)}
print(f"Set comprehension: {squares}")
```

## Set Characteristics


```python
print("\n=== Set Characteristics ===")

# 1. Unordered - no indexing
numbers = {3, 1, 4, 1, 5, 9, 2, 6}
print(f"Set: {numbers}")
```

numbers[0]  # TypeError! Sets don't support indexing
2. Unique elements only

```python
with_duplicates = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]
unique = set(with_duplicates)
print(f"Original list: {with_duplicates}")
print(f"As set (unique): {unique}")

# 3. Elements must be hashable (immutable)
valid_set = {1, "two", (3, 4)}  # int, str, tuple are hashable
# invalid_set = {1, [2, 3]}  # TypeError: unhashable type 'list'
```

## Membership Testing — O(1)


```python
print("\n=== Membership Testing ===")

valid_statuses = {"active", "pending", "completed", "failed"}

status = "active"
print(f"'{status}' in valid_statuses: {status in valid_statuses}")  # O(1)

status = "unknown"
print(f"'{status}' in valid_statuses: {status in valid_statuses}")  # O(1)

# Compare to list (O(n))
import time

large_set = set(range(1_000_000))
large_list = list(range(1_000_000))

search_value = 999_999

start = time.perf_counter()
_ = search_value in large_set
set_time = time.perf_counter() - start

start = time.perf_counter()
_ = search_value in large_list
list_time = time.perf_counter() - start

print(f"\nSet lookup: {set_time:.6f}s")
print(f"List lookup: {list_time:.6f}s")
print(f"Set is ~{list_time/set_time:.0f}x faster for membership testing")
```

## Set Operations


```python
print("\n=== Set Operations ===")

source_columns = {"id", "name", "email", "age"}
target_columns = {"id", "name", "phone", "address"}

print(f"Source columns: {source_columns}")
print(f"Target columns: {target_columns}")

# UNION (|) - All elements from both sets
all_columns = source_columns | target_columns  # or .union()
print(f"\nUnion (all columns): {all_columns}")

# INTERSECTION (&) - Elements in both sets
common_columns = source_columns & target_columns  # or .intersection()
print(f"Intersection (common): {common_columns}")

# DIFFERENCE (-) - Elements in first but not second
missing_in_target = source_columns - target_columns  # or .difference()
print(f"In source, not in target: {missing_in_target}")

extra_in_target = target_columns - source_columns
print(f"In target, not in source: {extra_in_target}")

# SYMMETRIC DIFFERENCE (^) - Elements in either, but not both
different_columns = source_columns ^ target_columns  # or .symmetric_difference()
print(f"Symmetric difference: {different_columns}")
```

## Set Comparisons


```python
print("\n=== Set Comparisons ===")

a = {1, 2, 3}
b = {1, 2, 3, 4, 5}
c = {1, 2, 3}

# Equality
print(f"{a} == {c}: {a == c}")  # True

# Subset
print(f"{a} <= {b}: {a <= b}")  # True (a is subset of b)
print(f"{a} < {b}: {a < b}")    # True (proper subset)
print(f"{a} < {c}: {a < c}")    # False (equal, not proper subset)

# Superset
print(f"{b} >= {a}: {b >= a}")  # True (b is superset of a)
print(f"{b} > {a}: {b > a}")    # True (proper superset)

# Disjoint (no common elements)
x = {1, 2, 3}
y = {4, 5, 6}
print(f"\n{x}.isdisjoint({y}): {x.isdisjoint(y)}")  # True
```

## Modifying Sets


```python
print("\n=== Modifying Sets ===")

columns = {"id", "name"}
print(f"Starting: {columns}")

# add() - Add single element
columns.add("email")
print(f"After add('email'): {columns}")

# update() - Add multiple elements
columns.update(["phone", "address"])
print(f"After update(['phone', 'address']): {columns}")

# remove() - Remove element (KeyError if not exists)
columns.remove("address")
print(f"After remove('address'): {columns}")

# discard() - Remove element (no error if not exists)
columns.discard("nonexistent")  # No error
print(f"After discard('nonexistent'): {columns}")

# pop() - Remove and return arbitrary element
popped = columns.pop()
print(f"pop() returned: {popped}, set now: {columns}")

# clear() - Remove all elements
columns.clear()
print(f"After clear(): {columns}")
```

## Frozenset — Immutable Set


```python
print("\n=== Frozenset ===")

# Frozenset is immutable - cannot be modified after creation
immutable_set = frozenset([1, 2, 3])
print(f"Frozenset: {immutable_set}")
```

Cannot add or remove
immutable_set.add(4)  # AttributeError
Can use as dict key (because it's hashable)

```python
config_by_features = {
    frozenset(["feature_a", "feature_b"]): "Config A",
    frozenset(["feature_c"]): "Config B",
}

features = frozenset(["feature_a", "feature_b"])
print(f"Config for {features}: {config_by_features[features]}")

# Can use in another set
set_of_sets = {frozenset([1, 2]), frozenset([3, 4])}
print(f"Set of frozensets: {set_of_sets}")

# Set operations work on frozensets too
fs1 = frozenset([1, 2, 3])
fs2 = frozenset([2, 3, 4])
print(f"\nfs1 | fs2: {fs1 | fs2}")  # Returns frozenset
print(f"fs1 & fs2: {fs1 & fs2}")
```

## Practical Applications


```python
print("\n=== Practical: Data Engineering Applications ===")

# 1. DEDUPLICATION
print("1. Deduplication:")
raw_ids = [101, 102, 101, 103, 102, 104, 101]
unique_ids = list(set(raw_ids))
print(f"  Original: {raw_ids}")
print(f"  Deduplicated: {unique_ids}")

# Preserve order while deduplicating (Python 3.7+)
def dedupe_preserve_order(items):
    """Remove duplicates while preserving order."""
    seen = set()
    result = []
    for item in items:
        if item not in seen:
            seen.add(item)
            result.append(item)
    return result

print(f"  Deduplicated (ordered): {dedupe_preserve_order(raw_ids)}")

# 2. SCHEMA COMPARISON
print("\n2. Schema Comparison:")
table_a_columns = {"id", "name", "email", "created_at"}
table_b_columns = {"id", "name", "phone", "updated_at"}

common = table_a_columns & table_b_columns
only_in_a = table_a_columns - table_b_columns
only_in_b = table_b_columns - table_a_columns

print(f"  Common columns: {common}")
print(f"  Only in Table A: {only_in_a}")
print(f"  Only in Table B: {only_in_b}")

# 3. VALIDATION
print("\n3. Validation:")
required_columns = {"id", "name", "email"}
actual_columns = {"id", "name", "phone"}

missing = required_columns - actual_columns
if missing:
    print(f"  Missing required columns: {missing}")
else:
    print("  All required columns present!")

# 4. FILTERING BY CATEGORY
print("\n4. Filtering by Category:")
valid_statuses = {"active", "pending"}
records = [
    {"id": 1, "status": "active"},
    {"id": 2, "status": "deleted"},
    {"id": 3, "status": "pending"},
    {"id": 4, "status": "unknown"},
]

valid_records = [r for r in records if r["status"] in valid_statuses]
print(f"  Records with valid status: {[r['id'] for r in valid_records]}")

# 5. UNIQUE VALUE EXTRACTION
print("\n5. Unique Value Extraction:")
records = [
    {"id": 1, "region": "NORTH"},
    {"id": 2, "region": "SOUTH"},
    {"id": 3, "region": "NORTH"},
    {"id": 4, "region": "EAST"},
]

unique_regions = {r["region"] for r in records}
print(f"  Unique regions: {unique_regions}")
```
