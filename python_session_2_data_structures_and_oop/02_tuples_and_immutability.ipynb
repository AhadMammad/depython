## Tuple Creation And Basics


```python
print("=== Tuple Creation and Basics ===")

# Tuple creation with parentheses
coordinates = (10, 20)
rgb_color = (255, 128, 0)
empty_tuple = ()

# Parentheses are optional (but recommended for clarity)
point = 10, 20, 30
print(f"point = {point}, type: {type(point)}")

# Single element tuple - needs trailing comma!
single = (42,)    # This is a tuple
not_tuple = (42)  # This is just an int!
print(f"single = {single}, type: {type(single)}")
print(f"not_tuple = {not_tuple}, type: {type(not_tuple)}")

# Tuple from iterable
chars = tuple("hello")
print(f"tuple('hello') = {chars}")
```

## Indexing And Slicing (Same As Lists)


```python
print("\n=== Indexing and Slicing ===")

data = ("id", "name", "email", "created_at", "updated_at")

print(f"data[0]:  '{data[0]}'")    # First element
print(f"data[-1]: '{data[-1]}'")   # Last element
print(f"data[1:3]: {data[1:3]}")   # Slice (returns tuple)
print(f"len(data): {len(data)}")

# Tuples support all read operations
print(f"'name' in data: {'name' in data}")
print(f"data.index('email'): {data.index('email')}")
print(f"data.count('name'): {data.count('name')}")
```

## Immutability — Tuples Cannot Be Modified


```python
print("\n=== Immutability ===")

record = (1, "Alice", "alice@example.com")
```

This would raise TypeError:
record[1] = "Bob"  # TypeError: 'tuple' object does not support item assignment
Cannot add or remove elements
record.append("new")  # AttributeError
BUT: If tuple contains mutable objects, those can be modified!

```python
mixed = (1, 2, [3, 4])
```

mixed[2].append(5)  # This works! The list inside is mutable
print(f"mixed after modifying inner list: {mixed}")
Create new tuple instead of modifying

```python
old_record = (1, "Alice", "active")
new_record = old_record[:2] + ("inactive",)  # Concatenation creates new tuple
print(f"Old: {old_record}")
print(f"New: {new_record}")
```

## Tuple Unpacking


```python
print("\n=== Tuple Unpacking ===")

# Basic unpacking
point = (10, 20, 30)
x, y, z = point
print(f"x={x}, y={y}, z={z}")

# Unpacking in function returns
def get_user_info():
    """Return multiple values as tuple."""
    return (1, "Alice", "alice@example.com")

user_id, name, email = get_user_info()
print(f"ID: {user_id}, Name: {name}, Email: {email}")

# Ignore values with underscore
_, name, _ = get_user_info()
print(f"Just the name: {name}")

# Extended unpacking with *
first, *middle, last = (1, 2, 3, 4, 5)
print(f"\nfirst={first}, middle={middle}, last={last}")
# OUTPUT: first=1, middle=[2, 3, 4], last=5

# Get first and rest
first, *rest = ("header", "row1", "row2", "row3")
print(f"first={first}, rest={rest}")

# Get all but last
*init, last = (1, 2, 3, 4, 5)
print(f"init={init}, last={last}")

# Swapping values (uses tuple packing/unpacking)
a, b = 10, 20
print(f"\nBefore swap: a={a}, b={b}")
a, b = b, a  # Tuple unpacking magic!
print(f"After swap: a={a}, b={b}")

# Unpacking in loops
records = [
    (1, "Alice", 100),
    (2, "Bob", 200),
    (3, "Charlie", 150),
]

print("\nUnpacking in loop:")
for user_id, name, score in records:
    print(f"  {name} (ID: {user_id}): {score} points")
```

## Named Tuples (Preview)


```python
print("\n=== Named Tuples ===")

# Regular tuple - positions are cryptic
user = (1, "Alice", "alice@example.com")
```

What is user[0]? user[1]? Hard to remember!
Named tuple - access by name

```python
User = namedtuple("User", ["id", "name", "email"])
user = User(id=1, name="Alice", email="alice@example.com")

print(f"Named tuple: {user}")
print(f"user.name: {user.name}")
print(f"user.email: {user.email}")
print(f"user[0]: {user[0]}")  # Still works with index

# Convert to dict
print(f"As dict: {user._asdict()}")

# Create from dict
user_dict = {"id": 2, "name": "Bob", "email": "bob@example.com"}
user2 = User(**user_dict)
print(f"From dict: {user2}")

# Replace values (creates new tuple)
user3 = user._replace(name="Alice Smith")
print(f"After replace: {user3}")

# Data engineering: Database row representation
Row = namedtuple("Row", ["customer_id", "order_id", "amount", "status"])
rows = [
    Row(101, 1001, 99.99, "completed"),
    Row(102, 1002, 149.50, "pending"),
    Row(101, 1003, 75.00, "completed"),
]

print("\nProcessing rows:")
for row in rows:
    print(f"  Customer {row.customer_id}: Order {row.order_id} - ${row.amount:.2f}")
```

## Why Immutability Matters


```python
print("\n=== Why Immutability Matters ===")

# 1. HASHABILITY — Tuples can be dict keys, lists cannot
print("1. Hashability:")

# Valid: tuple as dict key
location_data = {
    (40.7128, -74.0060): "New York",
    (51.5074, -0.1278): "London",
    (35.6762, 139.6503): "Tokyo",
}
print(f"  location_data[(40.7128, -74.0060)]: {location_data[(40.7128, -74.0060)]}")
```

Invalid: list as dict key
location_data[[40.7128, -74.0060]] = "NYC"  # TypeError: unhashable type: 'list'
2. THREAD SAFETY — Immutable objects are safe to share

```python
print("\n2. Thread Safety:")
print("  Tuples can be safely shared between threads without locks")
print("  Lists need synchronization if modified concurrently")

# 3. GUARANTEES — Value cannot change unexpectedly
print("\n3. Guarantees:")

def process_config(config):
    """Process configuration. Config is guaranteed not to change."""
    # With tuple, we know config won't be modified
    print(f"  Processing config: {config}")

config = ("localhost", 5432, "mydb")
process_config(config)
# config is guaranteed unchanged after function call
```

## Tuple Vs List — When To Use Which


```python
print("\n=== Tuple vs List — When to Use Which ===")

print("""
Use TUPLE when:
- Data has fixed structure (coordinates, RGB, database row)
- Data shouldn't change (configuration, constants)
- Need hashable type (dict key, set element)
- Multiple return values from function
- Unpacking a fixed number of values

Use LIST when:
- Collection of similar items (records, names)
- Need to add/remove items
- Order matters and may change
- Building up data iteratively

Examples:
  TUPLE: (lat, lng), (r, g, b), (x, y, z), ("John", 30, "NYC")
  LIST:  [user1, user2, user3], ["red", "blue", "green"]
""")
```

## Practical: Database Row Handling


```python
print("=== Practical: Database Row Handling ===")

# Define a schema using namedtuple
CustomerRow = namedtuple("CustomerRow", [
    "customer_id",
    "name",
    "email",
    "status",
    "created_at",
])

# Simulate database rows
db_rows = [
    (1, "Alice Smith", "alice@example.com", "active", "2024-01-15"),
    (2, "Bob Johnson", "bob@example.com", "inactive", "2024-01-16"),
    (3, "Charlie Brown", "charlie@example.com", "active", "2024-01-17"),
]

# Process rows with named access
active_customers = []
for row in db_rows:
    customer = CustomerRow(*row)  # Unpack tuple into named tuple
    if customer.status == "active":
        active_customers.append(customer)
        print(f"Active: {customer.name} ({customer.email})")

# Use as dict key for deduplication
seen_customers = {}
for row in db_rows:
    customer = CustomerRow(*row)
    # Use (name, email) tuple as unique key
    key = (customer.name, customer.email)
    if key not in seen_customers:
        seen_customers[key] = customer

print(f"\nUnique customers: {len(seen_customers)}")
```

## Summary


```python
print("\n=== Summary ===")
print("""
Tuple Basics:
- Immutable sequence: (1, 2, 3) or 1, 2, 3
- Single element: (x,) not (x)
- Supports indexing, slicing, iteration

Unpacking:
- Basic: a, b, c = tuple
- Extended: first, *rest = tuple
- Swap: a, b = b, a
- Loop: for x, y in [(1,2), (3,4)]

Named Tuples:
- namedtuple("Name", ["field1", "field2"])
- Access by name: obj.field1
- Still immutable and indexable

Why Immutability:
- Hashable (dict keys, set elements)
- Thread-safe (no synchronization needed)
- Guarantees (value won't change)

Tuple vs List:
- Tuple: Fixed structure, immutable data
- List: Growing collection, mutable data
""")
```
